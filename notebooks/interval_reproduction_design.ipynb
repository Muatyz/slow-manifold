# Interval Reproduction 设计

沿用 Dinc-style rank-2 vanilla CTRNN 与 `tanh` readout；不复刻 Ramesan 原网络。S1、S2、Go 为同一输入通道上的方波。

## Trial 与 loss

采样 $T\sim U[30,100)$、$T_{delay}\sim U[20,90)$，S1=50。Go 后等待 $T$，随后在等长的 response phase 输出 1。

$$L=\frac1B\sum_i[\lambda_{pre}\langle e^2\rangle_{pre}+\lambda_{wait}\langle e^2\rangle_{wait}+\lambda_{resp}\langle e^2\rangle_{resp}]$$

默认三个权重均为 1；每段先按时长归一化，再对 batch 平均。

## 评价与边界

预测时刻取 Go 后首次持续越过阈值的时间。记录 timing MAE/RMSE/bias、premature rate 与 no-response rate；无响应 trial 不进入 timing error，并由 no-response rate 单独报告。阈值准确率与 minimum rank 留待观察实验结果后定义。
