# 0. 省流

> 精确的连续吸引子虽然缺乏结构稳定性, 但是受到小扰动时, 并不会完全消失, 而是留下于原吸引子拓扑相同, 法向吸引而切向缓慢漂移的慢流形(近似连续吸引子). 
> 
> 因此, 连续吸引子在有限时间尺度(具体和行为有关)上, 具有功能鲁棒性. 

# 1. Fine-tune 问题


- 动力学: $\begin{aligned}\frac{\mathrm{d}\vec{x}}{\mathrm{d}t} = \vec{f}(\vec{x})\end{aligned}$
    - 状态变量: $\vec{x}\in\mathbb{R}^{d}$; 
- 不动点: $\forall\vec{x}\in\mathcal{M}$, $\vec{f}(\vec{x}) = \vec{0}$, $\mathcal{M}$ 为 $l$ 维的低维 ($l<d$) 流形
    - 线性化/Jacobian: 取 $\vec{x}^{*}\in\mathcal{M}$, 线性化方程 $\delta\dot{\vec{x}} = J(\vec{x}^{*})\delta\vec{x}$. 
        - Jacobian 特征值: 流形切向 $\lambda_{1} = \cdots = \lambda_{l} = 0$, 法向 $\mathrm{Re}(\lambda_{l+1}) = \cdots = \mathrm{Re}(\lambda_{d}) < 0$
- 微扰: 任意小的参数扰动使得 $\lambda_{\parallel} = 0$ 分裂为 $\lambda_{\parallel} = \pm \epsilon$




# 2. Bounded line attractor 模型

- 动力学: $\begin{aligned}\frac{\mathrm{d}\vec{x}}{\mathrm{d}t} = - \vec{x} + [\mathbf{W}\vec{x} + \vec{b}]_{+}\end{aligned}$
    - 权重矩阵 $\mathbf{W} = \begin{bmatrix}0 & -1 \\ -1 & 0 \end{bmatrix}$
    - 偏置 $\vec{b} = \begin{bmatrix}1 \\ 1 \end{bmatrix}$
- 不动点: 第一象限 ($x_{1},x_{2} > 0$) 内, 动力学 $\begin{cases} \dot{x}_{1} = -x_{1} -x_{2} + 1 \\ \dot{x}_{2} = -x_{1} - x_{2} + 1\end{cases}$, $\dot{x}_{1} = \dot{x}_{2} = 0$ 导出 $x_{1} + x_{2} - 1 = 0$. 
    - 不动点流形 $\mathcal{M}_{0} = \{(x_{1},x_{2}): x_{1} + x_{2} - 1 = 0, x_{1,2}\geq 0\}$
- 正则坐标化: $\begin{aligned}q_{\parallel} = \frac{x_{1}-x_{2}}{\sqrt{2}}\end{aligned}$, $\begin{aligned}q_{\perp} = \frac{x_{1}+x_{2}}{\sqrt{2}}\end{aligned}$
    - 动力学: $\begin{cases}\dot{q}_{\parallel} = 0 \\ \dot{q}_{\perp} = -2q_{\perp} \end{cases}\Rightarrow$ 特征值 $ \begin{cases}\lambda_{\parallel} = 0 \\ \lambda_{\perp} = -2 \end{cases}$
- 微扰 $\begin{cases}\mathbf{W} \to \mathbf{W} + \epsilon\mathbf{V}\\ \vec{b} \to \vec{b} + \epsilon\vec{c}\end{cases} \Rightarrow$ 正则动力学 $\begin{cases} \dot{q}_{\parallel} = \mathcal{O}(\epsilon) \\ \dot{q}_{\perp} = -2q_{\perp} + \mathcal{O}(\epsilon) \end{cases}\Rightarrow$ 时间尺度 $\begin{cases}\tau_{\parallel} \sim \epsilon^{-1} \\ \tau_{\perp} \sim 2^{-1} \end{cases}$
    - 微扰后的系统(单稳定不动点; 双稳定不动点+单不稳定不动点; 其它离散不动点组合...) 被称为原连续吸引流形的 `ghost`


# 3. Ring attractor 模型

> 果蝇中央复合体的 piecewise-linear ring model; Gaussian connectivity bump attractor; rank-2/low rank RNN; quasi-continuous attractor by EMPJ...

> $N = 6$ 的网络可能出现 $2,4,6,\cdots, 12$ 个不动点; 也有可能出现 limit cycle.

- $1$ 维环动力学 $\dot{\theta} = q(\theta)$ (状态变量 $\theta\in S^{1}$): 
    1. $q(\theta) = 0$ 无解: 系统为 limit cycle;
    2. $q(\theta) = 0$ 有解 $q(\theta^{*}) = 0$, 则 $\theta^{*}$ 为不动点
        - 稳定不动点和不稳定不动点必定交替出现 (因此通常为偶数个不动点)
        - 不同网络可能有不同数量的不动点, 但仍可视为同一连续吸引流形的不同近似

# 4. 微扰论

- 微扰动力学: $\dot{\vec{x}} = \vec{f}(\vec{x}) + \epsilon\vec{p}(\vec{x})$
    - 微扰尺度: $\epsilon\ll 1$, $||\vec{p}||_{\infty} = 1$
- 在 $l$ 维连续流形 $\mathcal{M}$ 附近, 将状态变量 $\vec{x}\in\mathbb{R}^{d}$ 拆分为 $\vec{x} = (\vec{q}_{\parallel},\vec{q}_{\perp})$
    - 流形切向坐标 $\vec{q}_{\parallel}\in\mathbb{R}^{l}$, 流形法向坐标 $\vec{q}_{\perp}\in\mathbb{R}^{d-l}$
- 快慢动力学分离: $\begin{cases} \dot{q}_{\parallel} = \epsilon\vec{g}(\vec{q}_{\parallel}, \vec{q}_{\perp}, \epsilon)\\ \dot{q}_{\perp} = \vec{h}(\vec{q}_{\parallel}, \vec{q}_{\perp}, \epsilon) \end{cases}$
    - $\epsilon = 0$ 时回归连续吸引流形 $\dot{\vec{y}} = 0$, 即 $\mathcal{M}_{0} = \{(\vec{q}_{\parallel}, \vec{q}_{\perp}) : \vec{h}(\vec{q}_{\parallel}, \vec{q}_{\perp}, 0) = 0\}$
- 正常双曲性 (normal hyperbolicity)
    - 法向 Jacobian $D_{q_{\perp}}\vec{h}$ 的本征值 $\lambda_{\perp}$ 实部严格为负(一致远离 $0$): $\mathrm{Re}(\lambda_{\perp}) < -\delta < 0$
    - 原连续吸引流形的切向本征值为 $0$: $\lambda_{\parallel} = 0$
- 持续流形定理(persistent manifold theorem)
    - $\epsilon\ll 1$, 原连续吸引流形 $\mathcal{M}_{0}\to\mathcal{M}_{\epsilon} = \{(\vec{q}_{\parallel},\vec{q}_{\perp}): \vec{q}_{\perp} = \vec{c}_{\epsilon}(\vec{q}_{\parallel})\}$
    - $\mathcal{M}_{\epsilon}$ 上的动力学 $\dot{q}_{\parallel} = \epsilon\vec{g}(\vec{q}_{\parallel}, \vec{c}_{\epsilon}(\vec{q}_{\parallel}), \epsilon)$, 切向速度为 $\mathcal{O}(\epsilon)$
    - 换元慢时间 $\tau=\epsilon t$, 动力学 $\begin{aligned}\frac{\mathrm{d}\vec{q}_{\parallel}}{\mathrm{d}\tau} = \vec{g}(\vec{q}_{\parallel}, \vec{c}_{\epsilon}(\vec{q}_{\parallel}), \epsilon)\end{aligned}$

# 5. 慢流形 $\mathcal{M}_{\epsilon}$ 如何偏离原连续吸引流形 $\mathcal{M}_{0}$

- $\mathcal{M}_{\epsilon}$ 上最大切向速度 $\eta = ||\dot{\vec{q}}_{\parallel}||_{\infty}$, $\exists \vec{p}_{\parallel}(\vec{x}) = -\vec{f}_{\parallel}(\vec{x})$ s.t. $\dot{\vec{q}}_{\parallel} = \vec{0}$
- 对于 `ghost`, 不必要求 $\lambda_{\parallel} = 0$, 只需要求 $|\lambda_{\parallel}|$ 充分小

# 6. 慢流形 $\mathcal{M}_{\epsilon}$ 的记忆误差

- 慢流形上的切向速度场 $\vec{\phi} = \vec{f}|_{\mathcal{M}}$
- 记忆误差 $\begin{aligned}\vec{x}(t) - \vec{x}_{0} = \int_{0}^{t}\phi[\vec{x}(s)]\mathrm{d}s\end{aligned}$
    - 误差上界 $\begin{aligned}e(t) = ||\vec{x}(t) - \vec{x}_{0}||\leq \int_{0}^{t}||\vec{\phi}[\vec{x}(s)]||\mathrm{d}s \leq t||\vec{\phi}||_{\infty} \ll \Delta\end{aligned}$
    - 只要满足容忍 $\Delta$, 则 $\mathcal{M}_{\epsilon}$ 在功能上和原连续吸引流形 $\mathcal{M}_{0}$ 无法区分
    - $t\to\infty$ 系统可能会丢失记忆; 有限 $T$ 误差可被容忍

# 7. 长期拓扑结构

- $e(t)\leq t||\vec{\phi}||_{\infty}$ 用于有限 $T$ 误差. 对于长时间, 系统会受到 $\mathcal{M}_{\epsilon}$ 上不动点或极限环的控制
    1. 不动点: $\theta_{0}\to\theta_{k}^{*}$. 若 $K$ 个稳定不动点均匀分布 $\begin{aligned}\left(\Delta\theta = \frac{2\pi}{K}\right)\end{aligned}$, 则最大误差 $\begin{aligned}e_{\max}\approx \frac{\Delta\theta}{2} = \frac{\pi}{K}\end{aligned}$. 不动点越多, 长期记忆越精细
    2. 极限环: 存储角度不断漂移. 长期来看, 记忆角度于当前输出之间的平均误差会趋于最大的无信息水平 (渐进误差为 $\pi$)

# 8. 实验设计

- vanilla RNN: 

    $$
    \begin{aligned}
    \vec{x}_{t} &= \varphi(\mathbf{W}_{\mathrm{in}}\vec{I}_{t} + \mathbf{W}\vec{x}_{t-1} + \vec{b}) + \vec{\zeta}_{t}, \quad \vec{\zeta}_{t} = \mathcal{N}(0,\sigma^{2}\mathbf{I}) \\
    \vec{y}_{t} &= \mathbf{W}_{\mathrm{out}}\vec{x}_{t} + \vec{b}_{\mathrm{out}}
    \end{aligned}
    $$

    - 动力学: $\mathrm{d}\vec{x} = [-\vec{x} + \varphi(\mathbf{W}_{\mathrm{in}}\vec{I} + \mathbf{W}\vec{x} + \vec{b})]\mathrm{d}t + \sigma\mathrm{d}\vec{W}_{t}$
    - Loss: masked MSE $\begin{aligned}\left\langle m_{i,t}\bigg(y_{i,t}-\hat{y}_{i,t}\bigg)^{2}\right\rangle_{i,t}\end{aligned}$
