# 0. 总纲

- 动力学: $\tau\dot{\vec{z}} = - \vec{z} + \mathbf{W}\vec{z} + \vec{x}(t)$
    - 神经活动状态矢量: $\vec{z}\in\mathbb{R}^{N}$
    - 递归连接矩阵: $\mathbf{W}\in\mathbb{R}^{N\times N}$
        - 连接权重分解: $\mathbf{W} = \mathbf{W}_{0} + \mathbf{W}_{1}$
            - $\mathbf{W}_{0} = \mathbf{U\Sigma V}^{\top}$: 强 low-rank structure
            - $\mathbf{W}_{1}$: 弱 full-rank random component
    - 外部输入/扰动: $\vec{x}(t)\in\mathbb{R}^{N}$

- 结论
    1. low-rank connectivity $\neq$ low-dim activity: high-dim $\vec{x}(t)$ 驱动产生的 response 仍然是 high-dim
    2. low-rank recurrent network 的反直觉现象(low-rank suppression): 越对齐 connectivity 的 low-rank modes, 反而越容易被压制
    3. response 的维度取决于 overlap matrix $\mathbf{P} = \mathbf{V}^{\top}\mathbf{U}$

# 1. Section 1: 反驳传统 RNN

- 传统 RNN: low-rank connectivity $\to$ low-dim recurrent computation $\to$ low-dim neural activity
    - Dinc: rank-2 $\mathbf{W} = \vec{m}^{(1)}\vec{n}^{(1)\top} + \vec{m}^{(2)}\vec{n}^{(2)\top}$, 对应 latent variables $(\kappa_{1},\kappa_{2}) = (\vec{n}^{(1)\top}\vec{x}(t), \vec{n}^{(2)\top}\vec{x}(t))$
    - 缺陷: external input 总是对齐 low-rank subspace, 而现实中的网络并非如此(sensory input, noise, perturbation...)
- 问题: strongly low-rank recurrent matrix + high-dim input $\to$ ? 

    严格讨论 connectivity dim, input dim 和 activity dim 的区别

# 2. Section 2: rank-1 network 产生 high-dim activity

- 动力学: $\tau\dot{\vec{z}} = (\mathbf{W}-\mathbf{I})\vec{z} + \vec{x}$. 约定 $\tau=1$
    - Jacobian $\mathbf{J} = \mathbf{W} - \mathbf{I}$: 对于 stable equilibrium 要求 autonomous dynamics $\mathrm{Re}[\lambda_{i}(\mathbf{W})] < 1$

- 构造 full-rank weight matrix $\mathbf{W} = \mathbf{W}_{0} + \mathbf{W}_{1}$
    - rank-1 recurrent matrix(strong) $\mathbf{W}_{0} = c\vec{u}\vec{u}^{\top}$, 其中 $||\vec{u}||=1$, $c=-10$ (取负值保证 stable)
    - full rank matrix(weak) $\begin{aligned}\mathbf{W}_{1} = \frac{\rho}{\sqrt{N}}\mathbf{Z}\end{aligned}$, 其中 $Z_{ij}\sim \mathcal{N}(0,1)$, $\rho = 0.5\ll c$
    - singular values: $\sigma_{1}(\mathbf{W})\approx |c|$ 特别大, 其余 $\sigma_{2,\cdots, N}\lesssim 2\rho$
