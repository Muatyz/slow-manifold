> 连续吸引子: $\vec{F}(\vec{x}) = \vec{0}$
> 
> slow/ghost set: $1\gg ||\vec{F}(\vec{x})||\neq 0$

# 1. 传统动力学

- 状态变量: 设 $N$ 个神经元, 活动矢量 $\vec{x}(t) = \begin{bmatrix} x_{1}(t) \\ x_{2}(t) \\ \vdots \\ x_{N}(t) \end{bmatrix} \in \mathbb{R}^{N}$
- 动力学: $\begin{aligned}\frac{\mathrm{d}\vec{x}}{\mathrm{d}t} = \vec{F}(\vec{x}; \vec{u}, \vec{\theta})\end{aligned}$
    - 变量
        - $\vec{x}$: 神经元活动矢量
        - $\vec{u}$: 外部输入
        - $\vec{\theta}$: 权重, 偏置等网络参数
        - $\vec{F}$: 速度矢量场. 
- 低维曲线 $\vec{x} = \vec{x}(s)$. 对于 HD 系统, $s = \theta\in [0,2\pi)$
    - 动力学分离: $\begin{aligned}\vec{F}(\vec{x}(s)) = \frac{\mathrm{d}\vec{x}}{\mathrm{d}s}\dot{s} + \vec{F}_{\perp}\end{aligned}$
        - $\vec{x}(s)$: 流形的几何形状
        - $\dot{s}$: 沿流形的速度
        - $\vec{F}_{\perp}$: 决定偏离流形后的吸引/排斥
    - 同样的轨迹 $\vec{x}(s)$ 可能由不同的动力学 ($\dot{s}$ 和 $\vec{F}_{\perp}$) 给出
- 不动点与稳定性(吸引子)
    - 不动点 $\vec{x}^{*}$ 满足 $\vec{F}(\vec{x}^{*}) = \vec{0}$
    - 稳定性: 不动点邻域 $\vec{x} = \vec{x}^{*} + \delta\vec{x}$, 一阶线性 $\begin{aligned}\frac{\mathrm{d}\delta\vec{x}}{\mathrm{d}t} = \mathbf{J}(\vec{x}^{*})\delta\vec{x}\end{aligned}$
        - Jacobian $\begin{aligned}\mathbf{J}(\vec{x}) = \frac{\partial\vec{F}}{\partial\vec{x}}\end{aligned}$, 本征方程 $\mathbf{J}(\vec{x}^{*})\hat{e}_{k} = \lambda_{k}\hat{e}_{k}$, 
        - $\begin{aligned}\delta\vec{x}(t) = \sum_{k} \frac{\delta\vec{x}(t)\cdot \hat{e}_{k}}{|\hat{e}_{k}|} \hat{e}_{k}\end{aligned}$
        - 动力学 $\begin{aligned}\frac{\mathrm{d}}{\mathrm{d}t}\sum_{k} \frac{\delta\vec{x}(t)\cdot \hat{e}_{k}}{|\hat{e}_{k}|} \hat{e}_{k} = \mathbf{J}(\vec{x}^{*})\sum_{k} \frac{\delta\vec{x}(t)\cdot \hat{e}_{k}}{|\hat{e}_{k}|} \hat{e}_{k} = \sum_{k} \lambda_{k} \frac{\delta\vec{x}(t)\cdot \hat{e}_{k}}{|\hat{e}_{k}|} \hat{e}_{k}\end{aligned}$; 分离 $\begin{aligned}\frac{\mathrm{d}\delta x_{k}(t)}{\mathrm{d}t} = \lambda_{k} \delta x_{k}(t)\end{aligned}$, 解得 $\delta x_{k}(t) = \delta x_{k}(0) e^{\lambda_{k} t}$
            - $\mathrm{Re}(\lambda_{k}) < 0$: 扰动衰减. $\lambda < 0$ 对应衰减时间 $\begin{aligned}\tau = \frac{1}{|\mathrm{Re}(\lambda_{k})|}\end{aligned}$, 越接近 $0$ 时间尺度越长
            - $\mathrm{Re}(\lambda_{k}) > 0$: 扰动增长
            - $\mathrm{Re}(\lambda_{k}) = 0$: 随遇平衡

# 2. 慢流形诊断

- 诊断量 $\begin{aligned}q(\vec{x}) = \frac{1}{2}||\vec{F}(\vec{x})||^{2}\end{aligned}$
    - $q(\vec{x}) = 0$: 经典不动点 ($\vec{F}(\vec{x}) = \vec{0}$)
    - $0< q(\vec{x})\ll 1$: $\vec{x}$ 被称作 slow point, 其集合则为 slow set. Ramesan 设定 $q <10^{-4}$. 
    - 性质: $\nabla q(\vec{x}) = [\mathbf{J}(\vec{x})]^{\top}\vec{F}(\vec{x})$

        > **证明**:
        > 
        > 状态变量 $\vec{x} = \begin{bmatrix} x_{1} \\ \ldots \\ x_{N} \end{bmatrix}$ , 速度场 $\vec{F}(\vec{x}) = \begin{bmatrix} F_{1}(\vec{x}) \\ \ldots \\ F_{N}(\vec{x}) \end{bmatrix}$, 诊断量 $\begin{aligned}q(\vec{x}) = \frac{1}{2}||\vec{F}(\vec{x})||^{2} = \frac{1}{2}[\vec{F}(\vec{x})]^{\top}\vec{F}(\vec{x}) = \frac{1}{2}\sum_{i=1}^{N}[F_{i}(\vec{x})]^{2}\end{aligned}$
        > 
        > Jacobian $\begin{aligned}\mathbf{J}(\vec{x}) = \frac{\partial\vec{F}}{\partial\vec{x}}\end{aligned}$, 矩阵元 $\begin{aligned}J_{ij}(\vec{x}) = \frac{\partial F_{i}}{\partial x_{j}}\end{aligned}$
        > 
        > $q$ 对 $x_{j}$ 的偏导 $\begin{aligned}\frac{\partial q}{\partial x_{j}} = \frac{\partial}{\partial x_{j}}\left\{\frac{1}{2}\sum_{i=1}^{N}[F_{i}(\vec{x})]^{2}\right\} = \sum_{i=1}^{N}F_{i}(\vec{x})\frac{\partial F_{i}(\vec{x})}{\partial x_{j}} = \sum_{i=1}^{N}J_{ij}F_{i} = [\mathbf{J}^{\top}\vec{F}]_{j}\end{aligned}$
        > 
        > 写作 $\nabla$ 形式: $\nabla q(\vec{x}) = [\mathbf{J}(\vec{x})]^{\top}\vec{F}(\vec{x})$

- near critical 的含义: 

- [Example] 动力学 $\dot{y} = F(y) = \mu + y^{2}$
    - $\mu<0$: 两不动点 $y_{\pm} = \pm\sqrt{-\mu}$
        - Jacobian $\begin{aligned}J(y) = \frac{\mathrm{d}F}{\mathrm{d}y} = \frac{\mathrm{d}}{\mathrm{d}y}(\mu+y^{2}) = 2y\Rightarrow \begin{cases} F(y_{+}) = +2\sqrt{-\mu} > 0 \\ F(y_{-}) = -2\sqrt{-\mu}<0\end{cases}\end{aligned}$, 则 $y_{+}$ 为不稳定不动点, $y_{-}$ 为稳定不动点
    - $\mu=0$: 两不动点在 $y=0$ 合并, Jacobian 本征值 $\lambda = 2y = 0$
    - $\mu>0$: 不动点消失. 在 $y\approx 0$ 附近, $\dot{y} = \mu+y^{2} \sim 0$
        - 穿过 $y\in [-L,L]$ 所需时间 $\begin{aligned}T = \int_{-L}^{L}\frac{\mathrm{d}y}{\dot{y}} = \int_{-L}^{L}\frac{\mathrm{d}y}{\mu + y^{2}} = \frac{2}{\sqrt{\mu}}\arctan{\frac{L}{\sqrt{\mu}}}\end{aligned}$
        - 极限 $\begin{aligned}\lim_{L\gg\sqrt{\mu}}T \approx \frac{\pi}{\sqrt{\mu}}\end{aligned}$; $\begin{aligned}\lim_{\mu\to 0^{+}}T\to\infty\end{aligned}$

# 3. 计算任务设定与神经网络

- Interval reproduction task
    - 三个输入脉冲: $S_{1}$, $S_{2}$, $\mathrm{Go}$; 时间间隔为 $T = t(S_{2}) - t(S_{1})$ 和 $T_{\mathrm{delay}} = t(\mathrm{Go}) - t(S_{2})$, 其中 $T_{\mathrm{delay}}$ 是可变的

        > 实验设定 $T\sim U[30,100)$, $T_{\mathrm{delay}}\sim U[20, 90)$

    - 目标: 收到 Go cue ($\mathrm{Go}$) 后, 等待 $T$ 时间间隔后产生输出

    - 网络要求: 
        1. 编码 $T$
        2. 在任意 $T_{\mathrm{delay}}$ 内保留 $T$
        3. 收到 $\mathrm{Go}$ 后复现 $T$
- vanilla RNN
    - 动力学: $\begin{aligned}\frac{\mathrm{d}\vec{h}}{\mathrm{d}t} = \vec{F}_{\vec{x}}(\vec{h}) = \frac{1}{\tau}\left[-\vec{h} + \mathbf{W}_{\mathrm{rec}}\tanh{(\vec{h})} + \mathbf{W}_{\mathrm{in}}\vec{x}\right]\end{aligned}$
    - 离散形式 $\begin{cases}\vec{h}_{t+1} = \vec{h}_{t} + \alpha[-\vec{h}_{t}+\mathbf{W}_{\mathrm{rec}}\tanh{(\vec{h}_{t})} + \mathbf{W}_{\mathrm{in}}\vec{x}_{t}] \\ \vec{y}_{t} = \mathbf{W}_{\mathrm{out}}\tanh{(\vec{h}_{t})}\end{cases}$, 其中 $\begin{aligned}\alpha = \frac{\mathrm{d}t}{\tau}\end{aligned}$
    - 对于固定 $\vec{x}_{0}$, 研究速度场 $\begin{aligned}\vec{F}_{\vec{x}_{0}}(\vec{h}) = \frac{1}{\tau}[-\vec{h} + \mathbf{W}_{\mathrm{rec}}\tanh{(\vec{h})} + \mathbf{W}_{\mathrm{in}}\vec{x}_{0}]\end{aligned}$
        - $\vec{x} = 0\to S_{1}\to 0$ 将会使整个速度场 $\vec{F}$ 都发生变化, 而非单独 $\vec{x}$ 受到 "推动"

- GRU
    - 动力学:
        - 非线性函数 $\begin{aligned}\sigma(x) = \frac{1}{1+e^{-x}}\end{aligned}$
        - 更新门(update gate) $\vec{z}_{t} = \sigma(\mathbf{W}_{z}\vec{x}_{t} + \mathbf{U}_{z}\vec{h}_{t-1} + \vec{b}_{z})$
        - 重置门(reset gate) $\vec{r}_{t} = \sigma(\mathbf{W}_{r}\vec{x}_{t} + \mathbf{U}_{r}\vec{h}_{t-1} + \vec{b}_{r})$
        - 候选隐状态 $\widetilde{\vec{h}}_{t} = \tanh{[\mathbf{W}_{h}\vec{x}_{t} + \mathbf{U}_{h}(\vec{r}_{t} \odot \vec{h}_{t-1}) + \vec{b}_{h}]}$

            > $[a_{1}, a_{2}, a_{3}] \odot [b_{1}, b_{2}, b_{3}] = [a_{1}b_{1}, a_{2}b_{2}, a_{3}b_{3}]$, 即逐元素乘法
        - 隐状态 $\vec{h}_{t} = (1 - \vec{z}_{t})\odot\vec{h}_{t-1} + \vec{z}_{t}\odot\widetilde{\vec{h}}_{t}$

            > $\vec{z}_{t}$ 很小时, 隐状态 $\vec{h}$ 更新很慢; $\vec{z}_{t}$ 很大时则迅速更新为候选隐状态
        - 输出 $\vec{y}_{t} = \tanh(\mathbf{W}_{\mathrm{out}}\vec{h}_{t} + \vec{b}_{\mathrm{out}})$

## Fig.5: 灵活时序计算的动力学系统模型

<img src="https://files.seeusercontent.com/2026/07/30/6goH/20260730121110245.png" width="70%">


构造 5 变量动力学系统 $(G_{1}, G_{2}, G_{3}, x, z)$

$$
\begin{aligned}
\dot{G}_{1} &= f(G_{1}) + \epsilon I(t) - G_{1}H(G_{3}-\theta),\quad H(a) = \begin{cases} 0, a<0 \\ 1, a>0\end{cases}\\
\dot{G}_{2} &= f(G_{2}) + \epsilon I(t)H(G_{1}-\theta) - G_{2}H(G_{3}-\theta)\\
\dot{G}_{3} &= f(G_{3}) + \epsilon I(t)H(G_{2}-\theta)\\
\dot{z} &= [H(G_{1}-\theta) - H(G_{2}-\theta)][1 - H(G_{3}-\theta)]\\
\dot{x} &= [r + (1 + \tanh{(x - K - \gamma)})^{2}] H(G_{3}-\theta) - H(\theta - G_{3})x, \quad K = r(z + 23.1)
\end{aligned}
$$

> $r = 0.02$, $\gamma = 1$, $\epsilon = 5$, $\theta = 4$

非线性函数 $f(u) = -0.5 (u - 2)^{3} + 2 (u - 2)^{2} - 4.7407$, 导数 $f^{\prime}(u) = -1.5(u-2)^{2} + 4(u-2)$. 

> 注意到 $f(u)$ 具有性质: 在 $\begin{aligned}u^{*} = \frac{14}{3}\end{aligned}$ 处有 $f^{\prime}(u^{*}) = 0$, $f(u^{*}) = 4.07\times 10^{-5} \approx 0$

模拟一遍任务过程: 

1. $S_{1}$ 刺激, 使得 $G_{1}$ 激活. $z$ 从 $S_{1}$ 开始线性积分时间

    $\begin{cases}H(G_{1}-\theta) = 1\\ H(G_{2}-\theta) = 0\\ H(G_{3}-\theta) = 0 \end{cases}\Rightarrow \dot{z} = (1-0)(1-0) = 1\Rightarrow z(t) = t - t_{S_{1}}$

2. $S_{2}$ 刺激, 使得 $G_{1}$ 仍高于阈值时, $G_{2}$ 被 $S_{2}$ 激活. $z$ 存储时间间隔 $T$

    $\begin{cases} H(G_{1}-\theta) = 1\\ H(G_{2}-\theta) = 1\\ H(G_{3}-\theta) = 0 \end{cases}\Rightarrow \dot{z} = (1-1)(1-0) = 0\Rightarrow z \approx T$

3. $S_{2}$ 后的延迟记忆阶段(delay). $G_{1}, G_{2}$ 仍高于阈值, $G_{3}$ 尚未激活. 

    $\begin{cases} H(G_{1}-\theta) = 1 \\ H(G_{2}-\theta) = 1 \\ H(G_{3}-\theta) = 0 \end{cases}\Rightarrow \dot{z} = (1-1)(1-0) = 0\Rightarrow z = z(t_{S_{2}})\approx T$

    而 $H(G_{3}-\theta) = 0$, $H(\theta-G_{3}) = 1$, 则 $\dot{x} = -x$, 使得 $x\approx 0$

4. $\mathrm{Go}$ 刺激, 使得 $G_{3}$ 激活. 

    $G_{2}$ 仍高于阈值: $H(G_{2}-\theta) = 1$, 则 $\dot{G}_{3} = f(G_{3}) + \epsilon I(t)H(G_{2}-\theta) = f(G_{3}) + \epsilon I(t)$ 开始增加 $G_{3}$. 

    $G_{3}$ 高于阈值后, $H(G_{3}-\theta) = 1$. 此时

    (1) $\begin{cases}\dot{G}_{1} = f(G_{1}) + \epsilon I(t) - G_{1} \\ \dot{G}_{2} = f(G_{2}) + \epsilon I(t)H(G_{1}-\theta) - G_{2}\end{cases}$ 在 $\mathrm{Go}$ 结束后 $(I(t)=0)$ 化为 $\begin{cases} \dot{G}_{1} = f(G_{1}) - G_{1} \\ \dot{G}_{2} = f(G_{2}) - G_{2} \end{cases}$, 负反馈使得 $G_{1,2}$ 拉回低活动状态, 即 reset

    (2) $1-H(G_{3}-\theta) = 0$ 使得 $\dot{z} = [H(G_{1}-\theta) - H(G_{2}-\theta)]\times 0 = 0 \Rightarrow z\approx T$, $\mathrm{Go}$ 使得 $z$ 变为控制 $x$ 动力学的固定参数

    (3) $H(G_{3}-\theta) = 1$, $H(\theta - G_{3}) = 0$ 使得 $x$ 的动力学从 $\dot{x} = -x$ 变为 $\dot{x} = r + [1 + \tanh{(x-r(z+23.1)-\gamma)}]^{2}$. 利用换元 $y = x - r(z+23.1) -\gamma$ 写作 $\dot{x} = r + [1 + \tanh{(y)}]^{2}$

5. $x$ 的慢速演化阶段

    reproduction 初期 $x\approx 0 \ll K + \gamma$, 则 $y\ll 0 \Rightarrow \tanh{(y)}\approx -1 \Rightarrow \dot{x} \approx r = 0.02$, $x$ 近似线性缓慢增加 $x(t) - x(t_{\mathrm{Go}}) \approx r(t - t_{\mathrm{Go}})$. 系统状态未停留在不动点, 而是以非零的小速度移动

    $z$ 通过 $K = r(z + 23.1) \approx r(T + 23.1)$ 控制 reproduction 的时间. 换元 $y_{\mathrm{out}} = x_{\mathrm{out}} - K - \gamma$ 或者说 $x_{\mathrm{out}} = y_{\mathrm{out}} + K + \gamma$
    
    慢速阶段 $x(t)\approx x_{0} + rt$, 则复现时间 $\begin{aligned}T_{\mathrm{rep}} \approx \frac{x_{\mathrm{out}} - x_{0}}{r} = \frac{K+\gamma + y_{\mathrm{out}} - x_{0}}{r} = z + 23.1 + \frac{\gamma + y_{\mathrm{out}} - x_{0}}{r}\end{aligned}$

    注意到 $\begin{aligned}\frac{\partial T_{\mathrm{rep}}}{\partial z} \approx 1\end{aligned}$, 即 $\Delta T_{\mathrm{rep}} \approx \Delta z$. 因此 $z\approx T\Rightarrow T_{\mathrm{rep}} \approx T$

6. $x$ 离开慢区并产生输出

    $x\uparrow$ 使得 $x-K-\gamma$ 从大负值增长到 $0$. 
    
    当 $x\approx K + \gamma$, $\tanh{(x-K-\gamma)}$ 有 $-1\to 0$, 则 $\dot{x} = r + (1+0)^{2} = r + 1\approx 100 r$, $x$ 迅速增加

    当 $x - K - \gamma\gg 0$, $\tanh{(x-K-\gamma)}$ 有 $0\to 1$, 则 $\dot{x} = r + (1+1)^{2} = r + 4\approx 400 r$, $x$ 更快地增长. 

    > 所以原图 Fig. 5b 的虚线存在一点问题, 不应该是保持在 1.

# 5. 一些谨慎观点

1. bifurcartion diagram 都是在 $N=3$ GRU 完成的, 高维系统难以追踪全部 FP 和 bifurcation. 

    高维网络的 slow set 尚未被证明都是 saddle node ghost

2. 作者通过 $q < \epsilon$ 得到 slow point, 再通过拟合得到 slow set. 不能证明是 slow manifold. 

3. $\lambda_{\max}$ 的作用有限. 

    non-normal Jacobian $\mathbf{J}\mathbf{J}^{\top}\neq \mathbf{J}^{\top}\mathbf{J}$. $\lambda_{\max}$ 不能完整决定扰动的短时间放大, eigenvector 方向

4. 5 变量模型是一个 toy model, 只是证明一种可能的机制, 不对 biological 负任何责任
